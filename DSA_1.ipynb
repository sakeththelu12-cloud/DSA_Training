# **2. Leaders in an Array**

```python
#nums = list(map(int, input().split()))
nums = [1, 3, 5, 7, 1, 2]
nums = [-3, 4, 6, 1, -3, -5]

ans = []

max = float('-inf')
for i in range(len(nums)-1,-1,-1):
    if max < nums[i]:
        max = nums[i]
        ans.append(nums[i])

ans.reverse()
print(ans)
```

# **3. Rearrange Array Elements by Sign**

```python
#nums = list(map(int, input().split()))
nums = [1, -2, -3, -4, 1, 3]
nums = [2, 3, 4, -1, -2, -3]

result = []
negetive, positive = [], []
for i in nums:
    if i < 0:
        negetive.append(i)
    else:
        positive.append(i)

for i,j in zip(positive,negetive):
    result.append(i)
    result.append(j)

print(result)
```

# **6. Kadane's Algorithm — Maximum Subarray Sum**

```python
#nums = list(map(int, input().split()))
nums = [-3,-4,-5,-6,-7,-10]
nums = [2, 3, 5, -2, 7, -4]

max_sum = float('-inf')
for i in range(len(nums)):
    sum_ = 0
    while i < len(nums):
        sum_ += nums[i]
        if max_sum < sum_:
            max_sum = sum_
        i += 1

print(max_sum)
```

```python
#nums = list(map(int, input().split()))
nums = [-2, -3, -7, -2, -10, -4]
nums = [2, 3, 5, -2, 7, -4]

current_sum = 0
for i in nums:
    current_sum = max(i, current_sum + i)
    max_sum = max(max_sum, current_sum)

print(max_sum)
```

# **7. Merge Two Sorted Arrays Without Extra Space**

```python
#nums1 = list(map(int, input().split()))
#nums2 = list(map(int, input().split()))
nums1 = [0, 2, 7, 8, 0, 0, 0]
nums2 = [-7, -3, -1]
nums1 = [-5, -2, 4, 5, 0, 0, 0]
nums2 = [-3, 1, 8]

m = len(nums1) - len(nums2)

for i in range(len(nums2)):
    for j in range(m + i):
        if nums2[i] <= nums1[j]:

            # Shift elements to the right
            for k in range(m + i, j, -1):
                nums1[k] = nums1[k - 1]

            # Insert the element
            nums1[j] = nums2[i]
            break

    else:
        nums1[m + i] = nums2[i]

print(nums1)


```

# **8. Count Inversions**

```python
#nums = list(map(int, input().split()))
nums = [-10, -5, 6, 11, 15, 17]
nums = [2, 3, 7, 1, 3, 5]

count = 0
for i in range(len(nums)):
    for j in range(i+1, len(nums)):
        if nums[j] < nums[i]:
            count += 1

print(count)
```

```python

def merge(left,right):
    global count
    i = 0
    j = 0
    result = []
    while i < len(left) and j < len(right):
        if left[i] > right[j]:
            count += len(left) - i
            result.append(right[j])
            j += 1
        else:
            result.append(left[i])
            i += 1

    result.extend(left[i:])
    result.extend(right[j:])
    return result

def mergeSort(nums):
    if len(nums) <= 1:
        return nums

    mid = len(nums)//2
    left = mergeSort(nums[:mid])
    right = mergeSort(nums[mid:])
    return merge(left,right)

    
count = 0

#nums = list(map(int, input().split()))
nums = [-10, -5, 6, 11, 15, 17]
nums = [2, 3, 7, 1, 3, 5]

print(mergeSort(nums))
print(count)
```
